```python
import Vigenere as vigenere
```

```python
# Example usage:
vigenere = Vigenere('secretkey')
plaintext = "This is a test message."
ciphertext = vigenere.Encrypt(plaintext)
print("Encrypted:", ciphertext)

decrypted_text = vigenere.Decrypt(ciphertext)
print("Decrypted:", decrypted_text)

# Encrypting and decrypting files
encrypted_content = vigenere.EncryptFile('example.txt')
print("Encrypted file content:", encrypted_content)

decrypted_content = vigenere.DecryptFile('example.txt')
print("Decrypted file content:", decrypted_content)
```
